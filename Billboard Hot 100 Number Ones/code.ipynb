```python
import pandas as pd
import numpy as np

artists = pd.read_csv("billboard_number_one_artists.csv")
songs = pd.read_csv("billboard_number_ones.csv")
```

```python
# Take the top 10 artists by total weeks spent at #1 (the true chart giants),
# then rank *those 10* by hits-per-week to see who got there efficiently vs. by camping on fewer songs.
top10_weeks = artists.nlargest(10, 'Total_Weeks_At_Number_One').copy()
top10_weeks['Hits_per_Week'] = top10_weeks['Number_Ones'] / top10_weeks['Total_Weeks_At_Number_One']
top10_weeks['Avg_Weeks_per_Hit'] = top10_weeks['Total_Weeks_At_Number_One'] / top10_weeks['Number_Ones']

top10_weeks[['Artist', 'Number_Ones', 'Total_Weeks_At_Number_One', 'Hits_per_Week', 'Avg_Weeks_per_Hit']] \
    .round(2).sort_values('Hits_per_Week', ascending=False).reset_index(drop=True)
```

### Takeaway

**Madonna** has the best ratio at **0.38** #1 hits per week (about 2.7 weeks per hit), with The Beatles close behind at 0.35 hits/week (about 2.8 weeks per hit). In contrast, Black Eyed Peas averaged just 0.11 hits/week (9.3 weeks at #1 per hit), showing they relied on fewer songs with much longer runs.
